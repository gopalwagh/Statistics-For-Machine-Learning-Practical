# EXPERIMENT 6: Implementation of Discriminant Function

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
class1 = np.array([
    [1, 1],
    [2, 1],
    [1, 2],
    [2, 2]
])

class2 = np.array([
    [5, 5],
    [6, 5],
    [5, 6],
    [6, 6]
])

mean1 = np.mean(class1, axis=0)
mean2 = np.mean(class2, axis=0)

W = mean2 - mean1
midpoint = (mean1 + mean2) / 2

bias = -np.dot(W, midpoint)

print("Class 1 Mean:", mean1)
print("Class 2 Mean:", mean2)
print("Weight Vector:", W)
print("Bias:", bias)

X_test = np.array([3, 3])
g = np.dot(W, X_test) + bias

print("\nTest Sample:", X_test)
print("Discriminant Value:", g)

if g >= 0:
    print("Predicted Class: Class 2")
else:
    print("Predicted Class: Class 1")

plt.scatter(
    class1[:, 0],
    class1[:, 1],
    label="Class 1",
    color="blue"
)
plt.scatter(
    class2[:, 0],
    class2[:, 1],
    label="Class 2",
    color="green"
)
plt.scatter(
    X_test[0],
    X_test[1],
    marker="x",
    s=100,
    color="red",
    label="Test Sample"
)

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Discriminant Function")
plt.legend()
plt.grid(True)

plt.show()
```
