# EXPERIMENT 8: Implementation of Parzen Window Density Estimation

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.array([
    2.1, 2.3, 2.5, 2.6,
    3.0, 3.2, 3.4, 3.7,
    4.0, 4.2, 4.5
])

h = 0.4

def gaussian_kernel(u):
    return (1 / np.sqrt(2 * np.pi)) * np.exp(-0.5 * u ** 2)

def parzen_estimate(x, samples, bandwidth):
    result = 0
    for sample in samples:
        u = (x - sample) / bandwidth
        result += gaussian_kernel(u)
    return result / (len(samples) * bandwidth)

x_values = np.linspace(1, 5.5, 300)

density = [
    parzen_estimate(x, data, h)
    for x in x_values
]

plt.plot(
    x_values,density,
    color="blue",
    linewidth=2,
    label="Parzen Density"
)

plt.hist(
    data,bins=8,
    density=True,alpha=0.3,
    color="orange",
    label="Histogram"
)

plt.xlabel("Data Value")
plt.ylabel("Probability Density")
plt.title("Parzen Window Density Estimation")

plt.legend()
plt.grid(True)

plt.show()
```
