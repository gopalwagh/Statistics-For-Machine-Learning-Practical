# EXPERIMENT 3: Implement Linear Regression algorithm

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python

X = np.array([[1], [2], [3], [4], [5], [6], [7]])
y = np.array([1.5, 2.3, 3.1, 4.0, 4.8, 5.6, 6.5])

# 2. Create and train the model
model = LinearRegression()
model.fit(X, y)

# 3. Get slope and intercept
slope = model.coef_[0]
intercept = model.intercept_

print(f"Model Equation: y = {slope:.2f}x + {intercept:.2f}")

# 4. Predict house price for 5.5 rooms
X_new = np.array([[5.5]])
y_pred = model.predict(X_new)

print(f"Predicted price for 5.5 rooms: ${y_pred[0] * 100:.2f}k")

# 5. Visualization

# Actual data points
plt.scatter(X, y, color='blue', label='Actual Data')

# Regression line
plt.plot(
    X,
    model.predict(X),
    color='red',
    linewidth=2,
    label='Regression Line'
)

plt.title('Simple Linear Regression')
plt.xlabel('Number of Rooms')
plt.ylabel('Price (in $100k)')
plt.legend()
plt.grid(True)

plt.show()
```
