## EXPERIMENT 4: Implementation of Bayes'

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report
```

```python
df = pd.read_csv("titanic.csv")
print(df.head())

features = ['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare']
target = 'Survived'

data = df[features + [target]].copy()
data['Age'] = data['Age'].fillna(data['Age'].median())
data['Sex'] = data['Sex'].map({
    'female': 0,
    'male': 1
})

data = data.dropna()

X = data[features]
y = data[target]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

model = GaussianNB()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print(f"Model Accuracy: {accuracy_score(y_test, y_pred) * 100:.2f}%\n")
print("Classification Report:")
print(classification_report(y_test, y_pred))

sample_passenger = pd.DataFrame(
    [[1, 0, 29.0, 0, 0, 100.0]],
    columns=features
)

survival_prob = model.predict_proba(sample_passenger)

print("\n--- Custom Passenger Prediction ---")
print(
    f"Probability of Not Surviving (Class 0): "
    f"{survival_prob[0][0] * 100:.2f}%"
)

print(
    f"Probability of Surviving (Class 1): "
    f"{survival_prob[0][1] * 100:.2f}%"
)
```
